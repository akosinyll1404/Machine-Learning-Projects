### Import the necesarry Library

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
import seaborn as sns
sns.set()

from sklearn.linear_model import LinearRegression
```

load the dataset

```python
filepath = '/content/drive/MyDrive/fuel_data.csv'
```

```python
data = pd.read_csv(filepath)
```

Explore the data

```python
data.head(7)
```

```python
data.rename(columns ={'Temperature (y)':'Temp', 'Pressure(x)':'Pressure'},inplace =True)
```

```python
data.head()
```

```python
data.describe()
```

```python
data.corr()
```

We will use the fuel rate and temperature for our linear regression

Test and Declare independent and dependent variable

```python
y = data['Fuel Rate']
x1 = data['Temp']
```

```python
plt.scatter(x1,y)
plt.xlabel('Temperature',fontsize=20)
plt.ylabel('Fuel Rate',fontsize=20)
plt.show()
```

```python
x = sm.add_constant(x1)
results = sm.OLS(y,x).fit()
results.summary()
```

```python
plt.scatter(x1,y)
yhat = x1*	-0.0317+24.6822
fig = plt.plot(x1,yhat, lw=4, c='orange', label ='regression line')
plt.xlabel('Temperature', fontsize = 20)
plt.ylabel('Fuel Rate', fontsize = 20)
plt.show()
```

Reshape the independent variable and Perform Linear Regression

```python
x_matrix = x1.values.reshape(-1,1)
```

```python
reg = LinearRegression()
reg.fit(x_matrix,y)
```

Calculate the R square

```python
reg.score(x_matrix,y)
```

Check  the intercept

```python
reg.intercept_
```

Check Coefficient

```python
reg.coef_
```

Predicted Values

```python
y_prediction = reg.predict(x_matrix)
y_prediction
```

Calculate MSE

```python
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(y,y_prediction)
mse
```

Calculate RMSE

```python
rmse = np.sqrt(mean_squared_error(y,y_prediction))
rmse
```
