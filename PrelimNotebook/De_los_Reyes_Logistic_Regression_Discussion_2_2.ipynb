Discussion 2.2 | Recitation - Logistic Regression

Suppose we had a logistic regression model with three features that learned the following bias and weights:
* b=1
* w1=3
* w2=−2
* w3=4

Further, suppose the following feature values for a given example:
* x1=0
* x2=5
* x3=1

Compute the logistic regression prediction for this particular example using python in Google Colab. Make sure to plot the solution.
Use the given reference for the computation and plotting of the result.

```python
# Lets list the variables
b = 1
w1 = 3
w2 = -2
w3 = 4
x1 = 0
x2 = 5
x3 = 1
```

```python
#Compute for the Z value
z = b + (w1 * x1) + (w2 * x2) + (w3 * x3)
z
```

```python
#Compute for the logistic regression prediction
import math
logistic_regression_prediction = 1 / (1 + math.exp(-z))
logistic_regression_prediction
```

## Plotting the Sigmoid function of Z value

```python
#import necessary library
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-10, 10, 100)
# Use NumPy's exp function to calculate y for each value in x
y = 1 / (1 + np.exp(-x))

plt.plot(x, y)
plt.plot(x, y)
# Scatter plot of z and corresponding y value
plt.scatter(z, 1 / (1 + np.exp(-z)), color='red')
plt.show()
```
